```python
import tkinter as tk
from tkinter import ttk


class ControlGUI:

    def __init__(self):
        self.root = tk.Tk()
        self.root.title("ALMA Observation Control")
        self.root.geometry("450x550")

        # -------------------------------------------------
        # Settings
        # -------------------------------------------------

        self.settings = {
            "ha_start": -3.0,
            "ha_end": 3.0,
            "declination": -30.0,
            "integration_time": 1.0,
            "frequency": 200.0,
            "image": "Galaxy",
            "vlbi": False
        }

        # -------------------------------------------------
        # GUI variables
        # -------------------------------------------------

        self.ha_start = tk.StringVar(value="-3 h")
        self.ha_end = tk.StringVar(value="+3 h")

        self.declination = tk.StringVar(value="-30°")

        self.integration_time = tk.StringVar(value="1 h")

        self.frequency = tk.StringVar(value="200 GHz")

        self.image = tk.StringVar(value="Galaxy")

        self.vlbi = tk.BooleanVar(value=False)

        self.running = False

        # -------------------------------------------------
        # Build GUI
        # -------------------------------------------------

        self.create_widgets()


    # =====================================================
    # GUI
    # =====================================================

    def create_widgets(self):

        title = ttk.Label(
            self.root,
            text="ALMA Observation Control",
            font=("Arial", 18)
        )

        title.pack(pady=20)


        # -------------------------------------------------
        # Observation
        # -------------------------------------------------

        observation_frame = ttk.LabelFrame(
            self.root,
            text="Observation"
        )

        observation_frame.pack(
            fill="x",
            padx=20,
            pady=10
        )


        # HA START

        ttk.Label(
            observation_frame,
            text="HA start:"
        ).grid(
            row=0,
            column=0,
            padx=10,
            pady=8,
            sticky="w"
        )

        self.ha_start_box = ttk.Combobox(
            observation_frame,
            textvariable=self.ha_start,
            values=[
                "-6 h",
                "-3 h",
                "0 h",
                "+3 h"
            ],
            state="readonly",
            width=15
        )

        self.ha_start_box.grid(
            row=0,
            column=1,
            padx=10
        )


        # HA END

        ttk.Label(
            observation_frame,
            text="HA end:"
        ).grid(
            row=1,
            column=0,
            padx=10,
            pady=8,
            sticky="w"
        )

        self.ha_end_box = ttk.Combobox(
            observation_frame,
            textvariable=self.ha_end,
            values=[
                "-3 h",
                "0 h",
                "+3 h",
                "+6 h"
            ],
            state="readonly",
            width=15
        )

        self.ha_end_box.grid(
            row=1,
            column=1,
            padx=10
        )


        # DECLINATION

        ttk.Label(
            observation_frame,
            text="Declination:"
        ).grid(
            row=2,
            column=0,
            padx=10,
            pady=8,
            sticky="w"
        )

        self.dec_box = ttk.Combobox(
            observation_frame,
            textvariable=self.declination,
            values=[
                "-80°",
                "-50°",
                "-30°",
                "-10°"
            ],
            state="readonly",
            width=15
        )

        self.dec_box.grid(
            row=2,
            column=1,
            padx=10
        )


        # INTEGRATION TIME

        ttk.Label(
            observation_frame,
            text="Integration time:"
        ).grid(
            row=3,
            column=0,
            padx=10,
            pady=8,
            sticky="w"
        )

        self.int_box = ttk.Combobox(
            observation_frame,
            textvariable=self.integration_time,
            values=[
                "0.2 h",
                "1 h",
                "50 h",
                "500 h"
            ],
            state="readonly",
            width=15
        )

        self.int_box.grid(
            row=3,
            column=1,
            padx=10
        )


        # FREQUENCY

        ttk.Label(
            observation_frame,
            text="Frequency:"
        ).grid(
            row=4,
            column=0,
            padx=10,
            pady=8,
            sticky="w"
        )

        self.freq_box = ttk.Combobox(
            observation_frame,
            textvariable=self.frequency,
            values=[
                "50 GHz",
                "100 GHz",
                "200 GHz",
                "250 GHz",
                "500 GHz"
            ],
            state="readonly",
            width=15
        )

        self.freq_box.grid(
            row=4,
            column=1,
            padx=10
        )


        # -------------------------------------------------
        # Sky model
        # -------------------------------------------------

        model_frame = ttk.LabelFrame(
            self.root,
            text="Sky Model"
        )

        model_frame.pack(
            fill="x",
            padx=20,
            pady=10
        )


        ttk.Label(
            model_frame,
            text="Model:"
        ).grid(
            row=0,
            column=0,
            padx=10,
            pady=10
        )


        self.model_box = ttk.Combobox(
            model_frame,
            textvariable=self.image,
            values=[
                "Galaxy",
                "Planet",
                "BH",
                "Misc1",
                "Misc2",
                "Misc3",
                "Misc4"
            ],
            state="readonly",
            width=15
        )

        self.model_box.grid(
            row=0,
            column=1,
            padx=10
        )


        # -------------------------------------------------
        # VLBI
        # -------------------------------------------------

        self.vlbi_check = ttk.Checkbutton(
            self.root,
            text="VLBI mode",
            variable=self.vlbi
        )

        self.vlbi_check.pack(
            pady=10
        )


        # -------------------------------------------------
        # Status
        # -------------------------------------------------

        self.status_label = ttk.Label(
            self.root,
            text="Status: Stopped"
        )

        self.status_label.pack(
            pady=10
        )


        # -------------------------------------------------
        # Buttons
        # -------------------------------------------------

        button_frame = ttk.Frame(self.root)

        button_frame.pack(
            pady=10
        )


        self.start_button = ttk.Button(
            button_frame,
            text="START",
            command=self.start
        )

        self.start_button.grid(
            row=0,
            column=0,
            padx=10
        )


        self.stop_button = ttk.Button(
            button_frame,
            text="STOP",
            command=self.stop
        )

        self.stop_button.grid(
            row=0,
            column=1,
            padx=10
        )


        self.apply_button = ttk.Button(
            self.root,
            text="Apply settings",
            command=self.apply_settings
        )

        self.apply_button.pack(
            pady=5
        )


    # =====================================================
    # Settings
    # =====================================================

    def apply_settings(self):

        # HA start
        self.settings["ha_start"] = float(
            self.ha_start.get().replace(" h", "")
        )

        # HA end
        self.settings["ha_end"] = float(
            self.ha_end.get().replace(" h", "")
        )

        # Declination
        self.settings["declination"] = float(
            self.declination.get().replace("°", "")
        )

        # Integration time
        self.settings["integration_time"] = float(
            self.integration_time.get().replace(" h", "")
        )

        # Frequency
        self.settings["frequency"] = float(
            self.frequency.get().replace(" GHz", "")
        )

        # Image
        self.settings["image"] = self.image.get()

        # VLBI
        self.settings["vlbi"] = self.vlbi.get()

        print("\nCurrent settings:")
        print(self.settings)


    # =====================================================
    # Start / Stop
    # =====================================================

    def start(self):

        self.apply_settings()

        self.running = True

        self.status_label.config(
            text="Status: Running"
        )

        print("Simulation started")


    def stop(self):

        self.running = False

        self.status_label.config(
            text="Status: Stopped"
        )

        print("Simulation stopped")


    # =====================================================
    # Main GUI loop
    # =====================================================

    def run(self):

        self.root.mainloop()
```
